```python
S1=pd.read_csv("/content/Table_S1.csv")
```

```python
S1.dropna(inplace=True)
```

```python
S1=S1.rename(columns={'SACs':'SAC','Ebin (eV)':'Ebin','ΔG*N2-end (eV)':'ΔG*N2-end','ΔG*N2-side (eV)':'ΔG*N2-side','ΔG*N2-N2H-end (eV)':'ΔG*N2-N2H-end','ΔG*N2-N2H-side (eV)':'ΔG*N2-N2H-side','ΔG*NH2-NH3 (eV)':'ΔG*NH2-NH3'})
```

```python
df_full=pd.merge(right=S1,left=S3,how='inner',on='SAC')
```

```python
df_full.info()
```

```python
df_end=df_full[['SAC','bond_length_end','Eads_N2-end','Eads_N2H-end','Eads_NH2','φ','Ebin','ΔG*N2-end','ΔG*N2-N2H-end',
       'ΔG*NH2-NH3']]
df_side=df_full[['SAC','bond_length_side','Eads_N2-side','Eads_N2H-side','Eads_NH2','φ','Ebin','ΔG*N2-side','ΔG*N2-N2H-side',
       'ΔG*NH2-NH3']]
```

```python
df_end['mode']=0
df_side['mode']=1
```

```python
df_end=df_end.rename(columns={'bond_length_end':'bond_length','Eads_N2-end':'Eads_N2','Eads_N2H-end':'Eads_N2H','ΔG*N2-end':'ΔG*N2','ΔG*N2-N2H-end':'ΔG*N2-N2H',})
```

```python
df_side=df_side.rename(columns={'bond_length_side':'bond_length','Eads_N2-side':'Eads_N2','Eads_N2H-side':'Eads_N2H','ΔG*N2-side':'ΔG*N2','ΔG*N2-N2H-side':'ΔG*N2-N2H',})
```

```python
df=pd.concat([df_end,df_side])
```

```python
df.sample(5)
```

```python
df.info()
```

```python
df["Metal"] = df["SAC"].str.split("@").str[0]
```

```python
df.sample(3)
```

```python
from mendeleev import element

def get_atomic_number(symbol):
    return element(symbol).atomic_number

def get_first_ionization_energy(symbol):
    return element(symbol).ionenergies.get(1)
```

```python
df["Atomic_Number"] = df["Metal"].apply(get_atomic_number)

df["First_Ionization_Energy_eV"] = df["Metal"].apply(
    get_first_ionization_energy
)
```

```python
df["Atomic_Radius"] = df["Metal"].apply(
    lambda symbol: element(symbol).atomic_radius
)

df["Electronegativity"] = df["Metal"].apply(
    lambda symbol: element(symbol).en_pauling
)

df["Valence_Electrons"] = df["Metal"].apply(
    lambda symbol: element(symbol).nvalence()
)
```

```python
intermediate_cols = [
    'ΔG*N2',
    'ΔG*N2-N2H',
    'ΔG*NH2-NH3'
]
df['UL'] = -df[intermediate_cols].max(axis=1)
```

```python
df.sample(5)
```

```python
df.to_csv('processed_data.csv')
```
